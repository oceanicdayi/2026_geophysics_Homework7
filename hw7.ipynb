```python
import numpy as np
```

## np.pi = 180
## np.pi/180 = 1

```python
fac = np.pi /180
```

```python
Ge = 978031.85
Go = 980717.39
lat = 48.1195
a1 = 0.005278895
a2 = 0.000023462
height = 487.9
```

```python
Gt = Ge * (1+a1*(np.sin(lat*fac)**2) + a2*(np.sin(lat*fac)**4)   )
```

```python
print("Gt = ", Gt)
```

```python
height_fac = 0.308
mass_fac = 0.112
```

```python
free_air_cor = height *  height_fac
print("free_air_cor : ", free_air_cor)

bouguer_cor = height *  mass_fac
print("bouguer_cor : ", bouguer_cor)
```

##　anomaly = observation - theoretical

```python
free_anomaly = ( Go + free_air_cor ) - Gt
print("free_anomaly: ", free_anomaly)
```

```python
bouguer_anomaly = ( Go + free_air_cor - bouguer_cor ) - Gt
print("bouguer_anomaly: ", bouguer_anomaly)
```
